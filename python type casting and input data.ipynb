```python
# Type Casting Examples

# 1. int to float
a = 10
b = float(a)
print("int to float:", b)  # Output: 10.0

# 2. float to int
x = 9.8
i = int(x)
print("float to int:", i)  # Output: 9

# 3. string to int
s = "123"
num = int(s)
print("string to int:", num)  # Output: 123

# 4. int to string
n = 456
text = str(n)
print("int to string:", text)  # Output: "456"

# 5. string to float
pi_str = "3.14"
pi = float(pi_str)
print("string to float:", pi)  # Output: 3.14

# 6. float to string
pi_text = str(pi)
print("float to string:", pi_text)  # Output: "3.14"
```
